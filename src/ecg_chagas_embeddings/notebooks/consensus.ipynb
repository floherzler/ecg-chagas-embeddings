## Spearman ranking agreement between models

We compute **Spearman’s ρ** between every pair of models using their **full test‑set prediction scores** (ranking vectors).  
This captures **global ranking consistency**: values near **1** mean two models largely agree on the ordering of samples; values near **0** indicate weak agreement; negative values indicate systematic disagreement.

The heatmap is symmetric with a 1.0 diagonal. Clusters indicate groups of models whose rankings are similar (e.g., same preprocessing regime or loss), which is what we want to diagnose.


```python
import altair as alt
import pandas as pd
from pathlib import Path

# Resolve repo root (look for configs/analysis/embeddings_probe_runs.toml)
root = Path.cwd().resolve()
marker = Path("configs/analysis/embeddings_probe_runs.toml")
if not (root / marker).exists():
    for parent in root.parents:
        if (parent / marker).exists():
            root = parent
            break
# Helper function to extract features for flexible sorting
def parse_model_name(name):
    parts = name.split("-")
    track = parts[0]
    exp_num = int(parts[1].replace("exp", ""))
    rot = "rot10" if parts[-1] == "rot10" else "none"
    loss_idx = -2 if rot == "rot10" else -1
    loss = parts[loss_idx]
    prepro = "-".join(parts[2:loss_idx])
    return pd.Series(
        [track, exp_num, prepro, loss, rot],
        index=["track", "exp_num", "prepro", "loss", "rot"],
    )
```

```python
spearman_path = root / "analysis/embeddings_probe/ranking_agreement/test/spearman_rho.csv"
source = pd.read_csv(spearman_path).rename(columns={"Unnamed: 0": "model_x"})

# Melt the wide correlation matrix into long format for Altair
source_long = source.melt(id_vars="model_x", var_name="model_y", value_name="spearman_rho")



# Apply parsing to both axes to allow consistent sorting
features_x = source_long['model_x'].apply(parse_model_name)
features_y = source_long['model_y'].apply(parse_model_name)
source_long = pd.concat([source_long, features_x.add_prefix('x_'), features_y.add_prefix('y_')], axis=1)

# Flexible sorting: Change the order of these columns to change heatmap grouping
sort_by = ['prepro','track','loss']
sort_cols_x = ['x_' + c for c in sort_by]
sort_cols_y = ['y_' + c for c in sort_by]

# Sort the dataframe to determine the axis order
model_order = source_long.sort_values(sort_cols_x)['model_x'].unique().tolist()

output = alt.Chart(source_long).mark_rect().encode(
    x=alt.X("model_x:O", title="Model X", sort=model_order),
    y=alt.Y("model_y:O", title="Model Y", sort=model_order),
    color=alt.Color("spearman_rho:Q", title="Spearman's ρ", scale=alt.Scale(scheme="redyellowgreen", domain=[-1, 1])),
    tooltip=[
        alt.Tooltip("model_x:O"),
        alt.Tooltip("model_y:O"),
        alt.Tooltip("spearman_rho:Q", format=".4f"),
        alt.Tooltip("x_prepro:N", title="Prepro X"),
        alt.Tooltip("x_loss:N", title="Loss X")
    ],
).properties(
    title="Spearman's ρ Global Ranking Agreement between Models",
    width=600,
    height=600,
)

output
```

```python
# --- Data Preparation ---
iou_path = root / "analysis/embeddings_probe/ranking_agreement/test/top5_iou.csv"
source_iou = pd.read_csv(iou_path).rename(columns={"Unnamed: 0": "model_x"}).melt(
    id_vars="model_x", var_name="model_y", value_name="top5_iou"
)

ktau_path = root / "analysis/embeddings_probe/ranking_agreement/test/top5_kendall_tau.csv"
source_ktau = pd.read_csv(ktau_path).rename(columns={"Unnamed: 0": "model_x"}).melt(
    id_vars="model_x", var_name="model_y", value_name="top5_kendall_tau"
)

# Sorting logic
sort_by = ["prepro", "track", "loss"]
features_x = source_iou["model_x"].apply(parse_model_name)
source_iou = pd.concat([source_iou, features_x.add_prefix("x_")], axis=1)
model_order = source_iou.sort_values(["x_" + c for c in sort_by])["model_x"].unique().tolist()

# --- Interactive Shared Selection ---
selection = alt.selection_point(fields=["model_x", "model_y"], on="mouseover", empty=False)

# --- Chart Construction ---
def make_chart(data, value_col, title, scheme, show_y=True):
    return (
        alt.Chart(data)
        .mark_rect()
        .encode(
            x=alt.X("model_x:O", title="Model X", sort=model_order),
            y=alt.Y("model_y:O", title="Model Y" if show_y else None, sort=model_order, axis=alt.Axis(labels=show_y, ticks=show_y)),
            color=alt.Color(f"{value_col}:Q", title=title, scale=alt.Scale(scheme=scheme, domain=[0, 1])),
            stroke=alt.condition(selection, alt.value("white"), alt.value(None)),
            strokeWidth=alt.condition(selection, alt.value(2), alt.value(0)),
            tooltip=["model_x:O", "model_y:O", alt.Tooltip(f"{value_col}:Q", format=".4f")],
        )
        .add_params(selection)
        .properties(title=title, width=400, height=400)
    )

chart_iou = make_chart(source_iou, "top5_iou", "Top 5 IoU", "inferno", show_y=True)
chart_ktau = make_chart(source_ktau, "top5_kendall_tau", "Top 5 Kendall Tau", "inferno", show_y=False)

# Combine and resolve Y-axis to ensure they are perfectly aligned and shared
(chart_iou | chart_ktau).resolve_scale(y="shared")
```
